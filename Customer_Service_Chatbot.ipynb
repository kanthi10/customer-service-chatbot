```python
print("Welcome to Customer Support!")
print("Hello! I am your customer service chatbot.")
print("Type 'bye' to exit.")

while True:
    question = input("How can I help you today? ").strip().lower()

    if question == "hello" or question == "hi" or question == "hey":
        print("Chatbot: Hello! How may I help you?")

    elif question == "refund":
        print("Chatbot: You can request a refund through our support team.")

    elif question == "orders" or question == "order":
        print("Chatbot: Please provide your order ID to check your order.")

    elif question == "hours":
        print("Chatbot: Our support team is available Monday to Saturday, 9 AM to 6 PM.")

    elif question == "contact":
        print("Chatbot: You can contact our support team at support@example.com.")

    elif question == "what can you do?":
          print("Chatbot: I can help with orders,refunds,support hours, and contact details.")

    elif question == "how can you help me?":
            print("Chatbot: I can answer common customer service questions.")

    elif question == "bye":
        print("Chatbot: Goodbye! Have a nice day.")
        break

    else:
        print("Chatbot: Sorry, I don't understand. Please contact customer support.")
```

Customer Service Chatbot

Project Overview

This project is a simple rule-based customer service chatbot developed using Python. It responds to common customer questions using predefined rules.

Features

- Responds to greetings such as hello, hi, and hey.
- Provides refund instructions.
- Answers order-related questions.
- Provides sample customer support hours and contact details.
- Responds to thanks and goodbye messages.
- Handles unknown questions with a default response.

Technologies Used

- Python
- Google Colab

How It Works

The chatbot accepts text input from the user and checks it against predefined conditions. It displays the corresponding response when a matching condition is found. The chatbot continues running until the user types "bye".

Testing

The chatbot was tested using greetings, refund requests, order inquiries, support-hour questions, contact inquiries, thank-you messages, and goodbye commands.

Limitations

This is a beginner-level, rule-based chatbot. It does not use a trained AI model, connect to a real customer database, or process actual orders and refunds.

Conclusion

This project demonstrates basic Python programming, conditional statements, loops, user input, and testing through a simple customer service application.
